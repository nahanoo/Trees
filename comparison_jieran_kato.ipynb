```python
from tree_jieran import Tree
import numpy as np
import time
import os.path as path
import pandas as pd
import os
```

```python
dataset_dir = "dataset"
tree_f = path.join(dataset_dir, "ENSG00000013016_EHD3_NT.table.dat")
seq_f = path.join(dataset_dir, "ENSG00000013016_EHD3_NT.msa.dat")
branch_f = path.join(dataset_dir, "ENSG00000013016_EHD3_NT.branchlength.dat")
```

```python
# Create the tree instance
tree_instance = Tree(tree_f, seq_f, branch_f)
u=0.3
Q = np.array([
    [-3*u, u, u, u],
    [u, -3*u, u, u],
    [u, u, -3*u, u],
    [u, u, u, -3*u]], dtype=np.float64)

# Time a single run (only the likelihood)
t0 = time.perf_counter()
ll = tree_instance.tree_loglikelihood(Q, use_P_cache=True)
t1 = time.perf_counter()

print("Log-likelihood:", ll)
print("Time (seconds):", t1 - t0)

# Optional: multiple runs for statistics
runs = 100 # Example Q matrix -- replace with your real model
times = []
for _ in range(runs):
    t0 = time.perf_counter()
    _ = tree_instance.tree_loglikelihood(Q, use_P_cache=True)
    times.append(time.perf_counter() - t0)

import numpy as _np
print("Mean:", _np.mean(times))
print("Std :", _np.std(times))
print("All :", times)
```

```python
# CSV file path
csv_file = "kato_jieran_likelihood_times.csv"

# Prepare new data
new_data = pd.DataFrame({
    "run_id": list(range(1, len(times)+1)),
    "run_time": times,
    "type": ["jieran"] * len(times)
})

# If file exists, append without writing the header
if os.path.exists(csv_file):
    new_data.to_csv(csv_file, mode="a", index=False, header=False)
else:
    # File doesn't exist, write header
    new_data.to_csv(csv_file, mode="w", index=False, header=True)

print(f"Appended {len(times)} runs to {csv_file}")
```
